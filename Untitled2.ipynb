```python
def constante_kaprekar(n):
    # Garante que o número tenha 4 dígitos preenchendo com zeros à esquerda
    n_str = f"{n:04d}"

    # Verifica se todos os dígitos são iguais (ex: 1111)
    if len(set(n_str)) == 1:
        return "Erro: Todos os dígitos são iguais. A subtração resultaria em 0 e não converge para 6174."

    iteracao = 0
    print(f"\n--- Iniciando com: {n_str} ---")

    # O loop continua até chegar na constante 6174
    # Nota: 0 também é um ponto de parada caso o número seja inválido,
    # mas o check acima já trata isso.
    while n != 6174 and n != 0:
        lista_digitos = list(n_str)
        ascendente = "".join(sorted(lista_digitos))
        descendente = "".join(sorted(lista_digitos, reverse=True))

        n = int(descendente) - int(ascendente)
        iteracao += 1

        n_str = f"{n:04d}"
        print(f"Passo {iteracao}: {descendente} - {ascendente} = {n_str}")

    return f"Sucesso! A constante 6174 foi atingida em {iteracao} passos."

# --- Interface com o Usuário ---
try:
    entrada = input("Digite um número de até 4 dígitos (com pelo menos dois diferentes): ")

    # Converte para inteiro para validar se é numérico
    numero_inicial = int(entrada)

    if 0 <= numero_inicial <= 9999:
        resultado = constante_kaprekar(numero_inicial)
        print(resultado)
    else:
        print("Erro: Por favor, digite um número entre 0 e 9999.")

except ValueError:
    print("Erro: Entrada inválida. Digite apenas números inteiros.")
```
