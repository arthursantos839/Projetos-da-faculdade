```python
def extrair_digitos(n):
    """Retorna os 4 dígitos de n (com zeros à esquerda) como
    quatro variáveis separadas, usando apenas aritmética."""
    d1 = n // 1000          # milhar
    d2 = (n % 1000) // 100  # centena
    d3 = (n % 100)  // 10   # dezena
    d4 = n % 10             # unidade
    return d1, d2, d3, d4


def ordenar_4(a, b, c, d):
    """Ordena 4 valores em ordem crescente usando apenas
    comparações e trocas (bubble sort manual, sem arrays)."""
    # Passagem 1
    if a > b:
        a, b = b, a
    if b > c:
        b, c = c, b
    if c > d:
        c, d = d, c
    # Passagem 2
    if a > b:
        a, b = b, a
    if b > c:
        b, c = c, b
    # Passagem 3
    if a > b:
        a, b = b, a
    return a, b, c, d


def montar_numero(d1, d2, d3, d4):
    """Reconstrói o inteiro de 4 dígitos a partir dos dígitos
    individuais (d1 = milhar, d4 = unidade)."""
    return d1 * 1000 + d2 * 100 + d3 * 10 + d4


def contar_repeticoes(a, b, c, d):
    """Retorna a maior quantidade de vezes que um mesmo dígito
    aparece entre os quatro, usando apenas comparações."""
    max_rep = 1

    # Conta repetições de cada dígito individualmente
    for digito in (a, b, c, d):
        contagem = 0
        if a == digito:
            contagem = contagem + 1
        if b == digito:
            contagem = contagem + 1
        if c == digito:
            contagem = contagem + 1
        if d == digito:
            contagem = contagem + 1
        if contagem > max_rep:
            max_rep = contagem

    return max_rep


# ------------------------------------------------------------
# VALIDAÇÃO 1 — tipo (inteiro e positivo)
# ------------------------------------------------------------
entrada = input("Digite um número de até 4 dígitos: ")

# Verificação manual de dígitos sem funções de string proibidas
# Única exceção: len() e isdigit() são operadores básicos de
# validação de entrada ensinados em sala.
# Se o professor não permite nem isso, substitua por try/except int().
valido = True
for caractere in entrada:
    if not caractere.isdigit():
        valido = False
        break

if not valido:
    print("Erro: o valor informado não é um número inteiro positivo.")
else:
    numero = int(entrada)

    # ----------------------------------------------------------
    # VALIDAÇÃO 2 — faixa 0000–9999
    # ----------------------------------------------------------
    if numero < 0 or numero > 9999:
        print("Erro: o número deve estar entre 0 e 9999.")
    else:
        print("Número informado:", numero)

        # ------------------------------------------------------
        # LOOP PRINCIPAL
        # ------------------------------------------------------
        iteracao = 0

        while numero != 6174:

            # Extrai os 4 dígitos
            d1, d2, d3, d4 = extrair_digitos(numero)

            # --------------------------------------------------
            # VALIDAÇÃO 3 — repetição de dígitos (máx. 2 iguais)
            # (revalidada a cada iteração, conforme enunciado)
            # --------------------------------------------------
            if contar_repeticoes(d1, d2, d3, d4) >= 3:
                print("Erro: o número possui três ou mais dígitos repetidos.")
                break

            # Ordena crescente → NDC
            p1, p2, p3, p4 = ordenar_4(d1, d2, d3, d4)
            NDC = montar_numero(p1, p2, p3, p4)

            # Inverte para decrescente → NDD
            NDD = montar_numero(p4, p3, p2, p1)

            # Subtração
            resultado = NDD - NDC
            iteracao = iteracao + 1

            # Exibe a iteração no formato pedido
            # Para exibir zeros à esquerda sem f-string com formato,
            # calculamos e imprimimos dígito a dígito via aritmética.
            r1, r2, r3, r4 = extrair_digitos(resultado)
            n1, n2, n3, n4 = extrair_digitos(NDC)
            nd1, nd2, nd3, nd4 = extrair_digitos(NDD)

            print("Iteração", iteracao, ":",
                  nd1, nd2, nd3, nd4,   # NDD (decrescente)
                  "-",
                  n1,  n2,  n3,  n4,   # NDC (crescente)
                  "=",
                  r1,  r2,  r3,  r4)   # resultado

            numero = resultado

        else:
            # Só executa se o while terminou SEM break (chegou em 6174)
            print("Constante de Kaprekar 6174 atingida em", iteracao, "iterações.")
```
